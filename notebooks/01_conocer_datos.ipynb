```python
import pandas as pd
carpeta = "../data/raw/"
```

```python
siniestros = pd.read_csv(carpeta + "Siniestros_Viales/SIGAT_ANUARIO_2024_Siniestros.csv.gz", low_memory=False)
print(siniestros.shape)
siniestros.head()
```

```python
siniestros.groupby("AA_Acc").agg(cantidad=("Codigo_Accidente", "count"))
```

```python
print(siniestros[["Latitud", "Longitud"]].isna().sum())
print(siniestros["Codigo_Accidente"].duplicated().sum())
```

Personas involucradas (Actor_vial)

```python
actores = pd.read_csv(carpeta + "Siniestros_viales/SIGAT_ANUARIO_2024_Actor_vial.csv.gz", low_memory=False)
print(actores.shape)
print(actores["Gravedad_Indicador_30d"].value_counts())
print(actores["Condicion"].value_counts())
```

Cámaras

Población
