# Cámaras Salvavidas
Objetivo: guardar la capa oficial de cámaras de la Secretaría de Movilidad (la misma del mapa de fotodeteccionbogota.com).
- URL: https://services2.arcgis.com/NEwhEo9GGSHXcRXV/ArcGIS/rest/services/Camaras_Salvavidas_5Feb2020/FeatureServer/2/query

```python
import requests
import pandas as pd

url = "https://services2.arcgis.com/NEwhEo9GGSHXcRXV/ArcGIS/rest/services/Camaras_Salvavidas_5Feb2020/FeatureServer/2/query"
parametros = {"where": "1=1", "outFields": "*", "f": "json"}

r = requests.get(url, params=parametros)
print(r.status_code)

datos = r.json()

filas = [f["attributes"] for f in datos["features"]]
camaras = pd.DataFrame(filas)

print(camaras.shape)
print(camaras["ELEMENTO"].value_counts()) 

camaras.to_csv("../data/raw/Camaras/camaras_sdm_arcgis.csv", index=False)
```
